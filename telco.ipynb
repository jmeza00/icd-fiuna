```python
import pandas as pd
from google.colab import files
url= "https://raw.githubusercontent.com/jmeza00/icd-fiuna/refs/heads/main/data/WA_Fn-UseC_-Telco-Customer-Churn.csv"
df= pd.read_csv(url)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

```

```python
display(df.head())

print("Dimensiones:", df.shape)

print("\nColumnas:")
print(df.columns.tolist())

print("\nTipos:")
print(df.dtypes)

print("\nInformación:")
df.info()
```

```python
faltantes = df.isna().sum()

print(faltantes[faltantes > 0])
```

```python
print(
    "Duplicados completos:",
    df.duplicated().sum()
)

print(
    "customerID duplicados:",
    df["customerID"].duplicated().sum()
)
```

```python
df["SeniorCitizen_Label"] = df["SeniorCitizen"].map({
    0: "No",
    1: "Yes"
})

df["Churn_Binario"] = df["Churn"].map({
    "No": 0,
    "Yes": 1
})

display(
    df[df["TotalCharges"].isna()]
)

display(
    df[
        df["TotalCharges"].isna()
    ][
        [
            "customerID",
            "tenure",
            "MonthlyCharges",
            "TotalCharges",
            "Churn"
        ]
    ]
)
```
